```python
import sncosmo
import numpy as np
import matplotlib.pyplot as plt
```

```python
#data = sncosmo.load_example_data()
```

## Constants

```python
ZP = 31.4
bandprefix = "lsst"
bandsnames = ["u","g","r","i","z","y"]
bandscolors = ["b","g","r","orange","grey","k"]
t = np.linspace(-20, 50, 100)  # Temps en jours autour du pic
NB = len(bandsnames)

z = 0.1
t0 = 0
x0 = 1.0
x1 = 0.0
c = 0.0
```

## Model

```python
# Charger un modèle de SN Ia (SALT2)
model = sncosmo.Model(source='salt2-extended')

# Définir les paramètres : redshift, temps du pic (t0), amplitude (x0), etc.
model.set(z=z, t0=t0 ,x0=x0, x1=x1, c=c)
```

## Fluxes

```python
fluxes = [ model.bandflux(bandprefix+bandsnames[ii],t,zpsys='ab',zp=ZP) for ii in range(NB) ]
```

## Plot

```python
fig,ax = plt.subplots(1,1)

for ii,flux in enumerate(fluxes):
    ax.plot(t,flux,color=bandscolors[ii],label=bandprefix+bandsnames[ii])
ax.legend()   

ax.set_xlabel("Time (days)")
ax.set_ylabel("Flux  (nJy) ")
ax.set_title(f"SNCOSMO (salt2) SNIA light curves in LSST at z= {z:.2f}")
plt.show()
```
